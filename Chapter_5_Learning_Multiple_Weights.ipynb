# Chapter 5 - Learning Multiple Weights at a Time: Generalizing Gradient Descent

## Tujuan Pembelajaran

Pada chapter ini kita belajar bagaimana neural network dapat menggunakan lebih dari satu weight untuk menghasilkan prediksi.

Pada chapter sebelumnya, kita hanya menggunakan satu weight. Sekarang, input dapat memiliki beberapa nilai dan setiap input mempunyai weight masing-masing.

Konsep utama yang dipelajari:

- Multiple inputs
- Multiple weights
- Vector
- Weighted sum
- Gradient descent untuk banyak weight
- Error
- Update weight

## 1. Dari Satu Weight ke Banyak Weight

Pada model sederhana, prediksi dapat dihitung dengan:

prediction = input × weight

Jika input yang digunakan lebih dari satu, maka setiap input memiliki weight sendiri.

Contohnya:

input = [0.8, 0.1, 0.9]

weight = [0.1, 0.3, 0.5]

Maka prediksi dihitung dengan menjumlahkan hasil perkalian setiap input dengan weight-nya:

prediction =
(0.8 × 0.1) +
(0.1 × 0.3) +
(0.9 × 0.5)

Dengan cara ini, neural network dapat mempertimbangkan beberapa informasi sekaligus.

```python
import numpy as np

# Input
inputs = np.array([0.8, 0.1, 0.9])

# Weight untuk setiap input
weights = np.array([0.1, 0.3, 0.5])

# Menghitung prediksi
prediction = np.dot(inputs, weights)

print("Prediction:", prediction)
```

## 2. Menghitung Error

Setelah mendapatkan prediction, kita membandingkannya dengan target.

Misalnya target = 1.

Error dapat dihitung menggunakan squared error:

error = (prediction - target)²

Semakin kecil nilai error, berarti prediction semakin dekat dengan target.

```python
target = 1

error = (prediction - target) ** 2

print("Prediction:", prediction)
print("Target:", target)
print("Error:", error)
```

## 3. Gradient untuk Banyak Weight

Karena terdapat beberapa weight, kita harus menghitung gradient untuk setiap weight.

Gradient menunjukkan arah perubahan weight agar error menjadi lebih kecil.

Untuk squared error:

gradient = 2 × input × (prediction - target)

Setiap weight memiliki gradient yang berbeda karena setiap input mempunyai nilai yang berbeda.

```python
# Menghitung selisih prediction dengan target
delta = prediction - target

# Menghitung gradient untuk setiap weight
gradient = 2 * inputs * delta

print("Delta:", delta)
print("Gradient:", gradient)
```

## 4. Updating Multiple Weights

Setelah gradient diperoleh, setiap weight diperbarui menggunakan learning rate.

Rumusnya:

weight = weight - learning_rate × gradient

Learning rate menentukan seberapa besar perubahan weight pada setiap proses pembelajaran.

```python
learning_rate = 0.1

# Update semua weight
weights = weights - learning_rate * gradient

print("Updated weights:", weights)
```

## 5. Melakukan Training Berulang

Neural network tidak langsung mendapatkan weight yang optimal dalam satu kali update.

Oleh karena itu, proses:

1. Menghitung prediction
2. Menghitung error
3. Menghitung gradient
4. Mengubah weight

dilakukan berulang kali.

Tujuannya adalah membuat error semakin kecil.

```python
inputs = np.array([0.8, 0.1, 0.9])
target = 1.0

# Weight awal
weights = np.array([0.1, 0.3, 0.5])

learning_rate = 0.1

for i in range(20):

    # Forward pass
    prediction = np.dot(inputs, weights)

    # Error
    error = (prediction - target) ** 2

    # Gradient
    delta = prediction - target
    gradient = 2 * inputs * delta

    # Update weights
    weights = weights - learning_rate * gradient

    print(
        f"Iterasi {i+1}: "
        f"Prediction = {prediction:.4f}, "
        f"Error = {error:.6f}"
    )

print("\nFinal weights:", weights)
```

## Kesimpulan Chapter 5

Pada chapter ini kita mengembangkan konsep gradient descent agar dapat menangani beberapa weight sekaligus.

Hal penting yang dipelajari:

- Satu input dapat memiliki satu weight.
- Banyak input membutuhkan banyak weight.
- Prediction dapat dihitung menggunakan dot product.
- Setiap weight memiliki gradient masing-masing.
- Semua weight dapat diperbarui secara bersamaan.
- Training dilakukan berulang kali sampai error menjadi lebih kecil.

Konsep ini menjadi dasar untuk membangun neural network dengan banyak neuron dan banyak layer.
