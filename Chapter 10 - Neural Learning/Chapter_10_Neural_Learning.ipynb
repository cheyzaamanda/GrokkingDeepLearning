## Learning Objectives

This chapter explains how a neural network learns by comparing its predictions with the correct targets and adjusting its weights.

The main objectives are:
1. Understand prediction error.
2. Calculate a simple loss value.
3. Update weights to reduce prediction error.
4. Understand the relationship between learning rate and weight updates.

## 1. How Neural Networks Learn

A neural network learns by making predictions, measuring prediction errors, and adjusting its weights.

The general learning process is:

1. Forward propagation: calculate a prediction from the input and current weights.
2. Loss calculation: measure the difference between the prediction and target.
3. Weight update: modify the weights to reduce the error.
4. Repeat the process over multiple iterations.

The learning process attempts to find weights that produce more accurate predictions.

```python
# Input and target
input_value = 1.0
target = 0.8

# Initial weight
weight = 0.2

# Calculate prediction
prediction = input_value * weight

# Calculate squared error
loss = (prediction - target) ** 2

print("Prediction:", prediction)
print("Target:", target)
print("Squared Error:", loss)
```

```python
# Reset the initial weight
weight = 0.2

# Set the learning rate
learning_rate = 0.1

# Calculate prediction
prediction = input_value * weight

# Calculate the gradient of squared error
gradient = 2 * (prediction - target) * input_value

# Update weight using gradient descent
weight = weight - learning_rate * gradient

# Calculate the new prediction and loss
new_prediction = input_value * weight
new_loss = (new_prediction - target) ** 2

print("Updated Weight:", weight)
print("New Prediction:", new_prediction)
print("New Loss:", new_loss)
```

```python
# Training settings
weight = 0.2
learning_rate = 0.1
epochs = 20

loss_history = []

for epoch in range(epochs):
    prediction = input_value * weight

    loss = (prediction - target) ** 2

    gradient = 2 * (prediction - target) * input_value

    weight = weight - learning_rate * gradient

    loss_history.append(loss)

    print(
        f"Epoch {epoch + 1:02d} | "
        f"Prediction: {prediction:.4f} | "
        f"Loss: {loss:.6f} | "
        f"Weight: {weight:.4f}"
    )
```

## 2. Gradient Descent

Gradient descent updates model parameters in the direction that reduces the loss.

The weight update rule is:

weight_new = weight_old - learning_rate × gradient

The gradient indicates how the loss changes when the weight changes.

The learning rate controls the size of each update. A learning rate that is too large may cause unstable updates, while one that is too small may make learning slow.

In the example, the gradient is calculated analytically for a single input and squared-error loss.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(range(1, epochs + 1), loss_history, marker="o")
plt.title("Loss During Training")
plt.xlabel("Epoch")
plt.ylabel("Squared Error")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Conclusion

Neural learning involves adjusting model weights to reduce prediction errors.

Gradient descent uses the loss gradient and learning rate to update weights. Repeated updates can gradually reduce the loss.

This simple example uses one input and one weight. Larger neural networks apply the same general principle to many weights and parameters.
