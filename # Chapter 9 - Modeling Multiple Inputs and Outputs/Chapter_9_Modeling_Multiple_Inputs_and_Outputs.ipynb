## 1. Modeling Multiple Inputs

A neural network can receive more than one input feature. Each input is multiplied by its corresponding weight, and the results are added together to produce a prediction.

For two inputs, the weighted sum is:

prediction = (input_1 × weight_1) + (input_2 × weight_2)

Weights determine how strongly each input contributes to the prediction.

This simple calculation is a foundation for more complex neural networks.

```python
# Define two input features
input_1 = 0.8
input_2 = 0.4

# Define their corresponding weights
weight_1 = 0.5
weight_2 = 0.7

# Calculate the prediction
prediction = (
    input_1 * weight_1
    + input_2 * weight_2
)

print("Input 1:", input_1)
print("Input 2:", input_2)
print("Weight 1:", weight_1)
print("Weight 2:", weight_2)
print("Prediction:", prediction)
```

```python
import numpy as np

# Input vector
inputs = np.array([0.8, 0.4, 0.6])

# Weight vector
weights = np.array([0.5, 0.7, 0.2])

# Calculate the weighted sum
prediction = np.dot(inputs, weights)

print("Inputs:", inputs)
print("Weights:", weights)
print("Prediction:", prediction)
```

## 2. Weighted Sum

The dot product calculates the sum of the products of corresponding elements.

For three inputs:

prediction = (input_1 × weight_1)
           + (input_2 × weight_2)
           + (input_3 × weight_3)

NumPy's `np.dot()` performs this calculation efficiently.

In a neural network, weights are learned from training data. They are adjusted during training so that predictions become closer to the expected targets.

```python
import numpy as np

# Three input features
inputs = np.array([0.8, 0.4, 0.6])

# Each row contains the weights for one output
weights = np.array([
    [0.5, 0.2, 0.1],
    [0.3, 0.7, 0.4]
])

# Calculate two outputs
outputs = np.dot(weights, inputs)

print("Inputs:", inputs)
print("Output 1:", outputs[0])
print("Output 2:", outputs[1])
print("All outputs:", outputs)
```

## 3. Modeling Multiple Outputs

A neural network can produce multiple outputs from the same input features.

Each output has its own set of weights. The first output uses the weights in the first row, while the second output uses the weights in the second row.

Matrix multiplication makes it possible to calculate multiple outputs efficiently.

The dimensions of the matrices must be compatible. In this example, the weight matrix has shape (2, 3), while the input vector has three elements. The result therefore contains two outputs.

## Conclusion

Multiple-input models combine information from several features using weights. Multiple-output models use separate weight sets to produce different predictions from the same input.

Weighted sums and matrix multiplication are essential building blocks for neural networks. These operations become especially important when a network contains many neurons and layers.
