## 2.1 What is Artificial Intelligence?

Artificial Intelligence (AI) adalah bidang ilmu komputer yang berfokus pada pembuatan sistem yang mampu melakukan tugas yang biasanya membutuhkan kecerdasan manusia.

Contohnya:
- Memahami bahasa manusia.
- Mengenali objek dalam gambar.
- Memberikan rekomendasi.
- Membantu mengambil keputusan.
- Mengenali suara.

AI merupakan konsep yang luas. Tidak semua sistem AI harus menggunakan Machine Learning atau Deep Learning.

## 2.2 What is Machine Learning?

Machine Learning (ML) adalah bagian dari AI yang memungkinkan komputer mempelajari pola dari data untuk membuat prediksi atau keputusan.

Pada pemrograman tradisional, manusia menuliskan aturan secara eksplisit.

Pada Machine Learning, sistem diberikan data dan algoritma untuk mempelajari pola yang dapat digunakan pada data baru.

**Contoh:**

Untuk membuat sistem prediksi harga rumah, kita dapat memberikan data seperti:
- Luas bangunan.
- Jumlah kamar.
- Lokasi.
- Harga rumah sebelumnya.

Model Machine Learning akan mempelajari hubungan antara fitur-fitur tersebut dengan harga rumah, kemudian menggunakan pola yang dipelajari untuk memprediksi harga rumah lainnya.

## 2.3 What is Deep Learning?

Deep Learning adalah bagian dari Machine Learning yang menggunakan Artificial Neural Network dengan banyak lapisan untuk mempelajari pola dari data.

Istilah "deep" mengacu pada banyaknya lapisan pemrosesan dalam neural network.

Deep Learning dapat mempelajari representasi data secara bertahap. Misalnya, pada pengenalan gambar:

1. Lapisan awal dapat mempelajari pola sederhana seperti garis dan tepi.
2. Lapisan berikutnya dapat mempelajari bentuk yang lebih kompleks.
3. Lapisan yang lebih dalam dapat menggabungkan pola tersebut untuk mengenali objek.

Kemampuan ini membuat Deep Learning berguna untuk menangani data yang kompleks seperti gambar, suara, dan bahasa.

## 2.4 Relationship Between AI, ML, and Deep Learning

AI, Machine Learning, dan Deep Learning memiliki hubungan bertingkat.

- AI merupakan bidang yang paling luas.
- Machine Learning merupakan salah satu pendekatan dalam AI.
- Deep Learning merupakan salah satu pendekatan dalam Machine Learning.

Dengan kata lain:

**Deep Learning ⊂ Machine Learning ⊂ Artificial Intelligence**

Artinya, setiap sistem Deep Learning termasuk Machine Learning dan AI, tetapi tidak semua sistem AI menggunakan Deep Learning.

## 2.5 Why Should We Learn Deep Learning?

Deep Learning penting dipelajari karena dapat membantu menyelesaikan berbagai permasalahan yang sulit ditangani menggunakan aturan pemrograman biasa.

Beberapa alasannya:

### 1. Automating Intelligence

Deep Learning dapat membantu mengotomatisasi tugas yang membutuhkan kemampuan mengenali pola, memahami informasi, atau membuat prediksi.

Contohnya adalah pengenalan wajah dan pengenalan suara.

### 2. Solving Complex Problems

Deep Learning mampu mempelajari pola kompleks dari data dalam jumlah besar.

Contohnya:
- Klasifikasi gambar.
- Penerjemahan bahasa.
- Analisis suara.
- Pemrosesan bahasa alami.

### 3. Creative Exploration

Deep Learning juga dapat digunakan untuk menghasilkan sesuatu, seperti teks, gambar, musik, dan konten lainnya.

### 4. Understanding Modern AI Systems

Dengan mempelajari Deep Learning dari dasar, kita dapat memahami bagaimana sistem AI bekerja, bukan hanya menggunakan library atau framework yang sudah tersedia.

## 2.6 Is Deep Learning Difficult to Learn?

Deep Learning memang memiliki konsep matematika dan pemrograman yang perlu dipahami. Namun, pembelajaran dapat dilakukan secara bertahap.

Beberapa konsep dasar yang akan membantu:

- **Python:** digunakan untuk menulis implementasi.
- **Mathematics:** terutama operasi dasar, aljabar, dan konsep turunan.
- **Linear Algebra:** membantu memahami data dan operasi pada neural network.
- **Problem Solving:** membantu memahami bagaimana model belajar dari kesalahan.

Tidak harus langsung memahami seluruh konsep matematika secara mendalam. Pemahaman dapat dibangun melalui contoh sederhana dan eksperimen kode.

## 2.7 Learning Deep Learning from Scratch

Salah satu pendekatan untuk mempelajari Deep Learning adalah memahami proses pembentukan model dari dasar.

Daripada hanya menggunakan framework yang sudah menyediakan fungsi neural network, kita dapat mempelajari komponen dasarnya terlebih dahulu.

Tahapan pembelajaran secara umum:

1. Memahami bagaimana model menghasilkan prediksi.
2. Memahami bagaimana model mengukur kesalahan prediksi.
3. Memahami bagaimana bobot model diperbarui.
4. Mempelajari bagaimana beberapa bobot dan neuron bekerja bersama.
5. Memahami bagaimana neural network belajar melalui backpropagation.

Dengan memahami proses tersebut, penggunaan framework seperti TensorFlow dan PyTorch akan menjadi lebih mudah dipahami.
