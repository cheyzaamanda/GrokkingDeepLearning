## 2.1 What is Gradient Descent?

Gradient Descent merupakan algoritma optimasi yang digunakan untuk mencari nilai parameter model agar menghasilkan error yang lebih kecil.

Dalam neural network, parameter yang diperbarui salah satunya adalah weight.

Prosesnya:

1. Model menghasilkan prediction.
2. Prediction dibandingkan dengan target.
3. Model menghitung loss.
4. Gradient dihitung untuk mengetahui arah perubahan weight.
5. Weight diperbarui.
6. Proses diulang sampai loss menjadi lebih kecil.

Tujuan utamanya adalah menemukan nilai weight yang dapat meminimalkan loss.

## 2.2 Understanding Loss

Loss merupakan ukuran seberapa jauh hasil prediksi model dari target sebenarnya.

Contoh:

Target = 10

Prediction = 6

Error = 10 - 6 = 4

Salah satu fungsi loss yang sederhana adalah Squared Error.

Rumus:

Loss = (Prediction - Target)²

Contoh:

Loss = (6 - 10)²

Loss = 16

Squared Error membuat kesalahan bernilai positif dan memberikan penalti lebih besar pada kesalahan yang besar.

## 2.3 Understanding Gradient

Gradient menunjukkan arah dan seberapa besar perubahan suatu fungsi terhadap parameter.

Dalam Gradient Descent, gradient digunakan untuk menentukan bagaimana weight harus diperbarui agar loss berkurang.

Secara sederhana:

- Jika perubahan weight membuat loss meningkat, arah perubahan perlu diperbaiki.
- Jika perubahan weight membuat loss menurun, arah tersebut membantu model belajar.

Gradient Descent bergerak ke arah yang berlawanan dengan gradient untuk menurunkan nilai loss.

## 2.4 Learning Rate

Learning Rate merupakan nilai yang mengatur seberapa besar perubahan weight pada setiap proses training.

Rumus umum:

New Weight = Old Weight - (Learning Rate × Gradient)

Learning Rate memiliki pengaruh terhadap proses pembelajaran.

### Learning Rate terlalu kecil

- Perubahan weight sangat sedikit.
- Training dapat berlangsung lebih lambat.

### Learning Rate terlalu besar

- Perubahan weight terlalu besar.
- Model dapat melewati nilai optimal.
- Loss dapat menjadi tidak stabil.

### Learning Rate yang sesuai

Membantu model memperbarui weight secara bertahap menuju nilai yang menghasilkan loss lebih kecil.

## 2.5 Gradient Descent in a Simple Neural Network

Misalkan sebuah model memiliki satu input dan satu weight.

Rumus prediction:

Prediction = Input × Weight

Loss:

Loss = (Prediction - Target)²

Gradient terhadap weight:

Gradient = 2 × Input × (Prediction - Target)

Weight diperbarui menggunakan:

New Weight = Old Weight - (Learning Rate × Gradient)

Dengan rumus tersebut, model dapat memperbarui weight berdasarkan besar dan arah kesalahan prediksi.

Proses ini akan dilakukan berulang kali selama training.

## 2.5 Gradient Descent in a Simple Neural Network

Misalkan sebuah model memiliki satu input dan satu weight.

Rumus prediction:

Prediction = Input × Weight

Loss:

Loss = (Prediction - Target)²

Gradient terhadap weight:

Gradient = 2 × Input × (Prediction - Target)

Weight diperbarui menggunakan:

Dengan rumus tersebut, model dapat memperbarui weight berdasarkan besar dan arah kesalahan prediksi.

Proses ini akan dilakukan berulang kali selama training.

```python
# Nilai target sebenarnya
target = 10

# Hasil prediksi model
prediction = 6

# Menghitung squared error
loss = (prediction - target) ** 2

print("Target:", target)
print("Prediction:", prediction)
print("Loss:", loss)
```

## 3.2 Code Explanation

Penjelasan:

- `target` merupakan nilai sebenarnya.
- `prediction` merupakan hasil yang diprediksi model.
- `prediction - target` menghitung selisih prediksi dengan target.
- `** 2` digunakan untuk mengkuadratkan error.

Hasil loss menunjukkan seberapa besar kesalahan model.

Semakin kecil nilai loss, semakin dekat prediction dengan target.

## 4.1 Updating Weight Using Gradient Descent

Sekarang kita akan memperbarui weight menggunakan Gradient Descent.

Model menggunakan satu input dan satu weight.

Rumus:

Prediction = Input × Weight

Gradient = 2 × Input × (Prediction - Target)

New Weight = Old Weight - (Learning Rate × Gradient)

Pada contoh ini, weight akan diperbarui satu kali berdasarkan loss yang dihasilkan.

```python
# Input dan target
input_value = 2
target = 10

# Weight awal
weight = 3

# Learning rate
learning_rate = 0.01

# Forward propagation
prediction = input_value * weight

# Menghitung gradient
gradient = 2 * input_value * (prediction - target)

# Memperbarui weight
new_weight = weight - (learning_rate * gradient)

# Menghitung loss sebelum dan sesudah update
loss_before = (prediction - target) ** 2

new_prediction = input_value * new_weight
loss_after = (new_prediction - target) ** 2

print("Old Weight:", weight)
print("Prediction Before:", prediction)
print("Loss Before:", loss_before)

print("\nGradient:", gradient)
print("New Weight:", new_weight)

print("\nPrediction After:", new_prediction)
print("Loss After:", loss_after)
```

## 4.2 Code Explanation

Penjelasan kode:

1. `input_value` merupakan input yang diberikan kepada model.
2. `target` merupakan nilai yang ingin diprediksi.
3. `weight` merupakan bobot awal model.
4. `learning_rate` menentukan besar perubahan weight.
5. `prediction` menghitung hasil prediksi menggunakan Forward Propagation.
6. `gradient` menghitung arah dan besar perubahan loss terhadap weight.
7. `new_weight` memperbarui weight menggunakan Gradient Descent.
8. `loss_before` dan `loss_after` digunakan untuk membandingkan kesalahan sebelum dan sesudah pembaruan.

Jika proses berjalan dengan baik, loss setelah pembaruan weight akan lebih kecil daripada loss sebelumnya.

## 5.1 Training Through Multiple Iterations

Dalam proses training, pembaruan weight tidak hanya dilakukan satu kali.

Model akan melakukan proses yang sama secara berulang.

Setiap iterasi terdiri dari:
1. Menghasilkan prediction.
2. Menghitung loss.
3. Menghitung gradient.
4. Memperbarui weight.

Dengan melakukan pembaruan berulang, model diharapkan dapat menemukan weight yang menghasilkan loss lebih kecil.

```python
# Input dan target
input_value = 2
target = 10

# Weight awal
weight = 3

# Learning rate
learning_rate = 0.01

# Jumlah iterasi
iterations = 20

# Training loop
for i in range(iterations):

    # Forward propagation
    prediction = input_value * weight

    # Menghitung loss
    loss = (prediction - target) ** 2

    # Menghitung gradient
    gradient = 2 * input_value * (prediction - target)

    # Memperbarui weight
    weight = weight - (learning_rate * gradient)

    print(
        f"Iteration {i + 1}: "
        f"Weight = {weight:.4f}, "
        f"Prediction = {prediction:.4f}, "
        f"Loss = {loss:.4f}"
    )
```

## 5.2 Code Explanation

Pada kode tersebut, proses training dilakukan menggunakan perulangan `for`.

Setiap iterasi:
- Model menghitung prediction berdasarkan weight saat ini.
- Loss dihitung untuk mengukur kesalahan.
- Gradient digunakan untuk menentukan perubahan weight.
- Weight diperbarui menggunakan learning rate.

Setelah weight diperbarui, iterasi berikutnya menggunakan weight yang baru.

Dengan demikian, model belajar secara bertahap dari kesalahan sebelumnya.

## 5.2 Code Explanation

Pada kode tersebut, proses training dilakukan menggunakan perulangan `for`.

Setiap iterasi:
- Model menghitung prediction berdasarkan weight saat ini.
- Loss dihitung untuk mengukur kesalahan.
- Gradient digunakan untuk menentukan perubahan weight.
- Weight diperbarui menggunakan learning rate.

Setelah weight diperbarui, iterasi berikutnya menggunakan weight yang baru.

Dengan demikian, model belajar secara bertahap dari kesalahan sebelumnya.

```python
import matplotlib.pyplot as plt

# Input dan target
input_value = 2
target = 10

# Weight awal
weight = 3

# Learning rate
learning_rate = 0.01

# Jumlah iterasi
iterations = 30

# Menyimpan loss
loss_history = []

for i in range(iterations):

    # Forward propagation
    prediction = input_value * weight

    # Menghitung loss
    loss = (prediction - target) ** 2

    # Menyimpan loss
    loss_history.append(loss)

    # Menghitung gradient
    gradient = 2 * input_value * (prediction - target)

    # Memperbarui weight
    weight = weight - learning_rate * gradient

# Visualisasi loss
plt.plot(loss_history, marker="o")

plt.title("Loss During Gradient Descent")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.grid(True)

plt.show()
```

## 7.1 Effect of Learning Rate

Learning Rate memengaruhi seberapa cepat weight berubah selama training.

Untuk memahami pengaruhnya, kita akan mencoba beberapa nilai learning rate:

- 0.001
- 0.01
- 0.1

Setiap percobaan menggunakan input, target, dan weight awal yang sama.

Tujuannya adalah melihat bagaimana learning rate memengaruhi penurunan loss.

```python
import matplotlib.pyplot as plt

input_value = 2
target = 10

learning_rates = [0.001, 0.01, 0.1]
iterations = 30

for lr in learning_rates:

    weight = 3
    loss_history = []

    for i in range(iterations):

        prediction = input_value * weight

        loss = (prediction - target) ** 2
        loss_history.append(loss)

        gradient = 2 * input_value * (prediction - target)

        weight = weight - lr * gradient

    plt.plot(
        loss_history,
        label=f"Learning Rate = {lr}"
    )

plt.title("Comparison of Learning Rates")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```
