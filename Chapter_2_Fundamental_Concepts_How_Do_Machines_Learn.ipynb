## 2.1 How Do Machines Learn?

Mesin belajar dengan menemukan hubungan atau pola dari data.

Contohnya, kita ingin memprediksi apakah seseorang menyukai sebuah film berdasarkan beberapa informasi.

Model diberikan:
- Input: informasi film.
- Output: apakah film disukai atau tidak.

Model akan mempelajari hubungan antara input dan output tersebut.

Proses pembelajaran dilakukan dengan menyesuaikan parameter model agar hasil prediksi semakin mendekati target sebenarnya.

## 2.2 Prediction

Prediction adalah proses ketika model menghasilkan perkiraan output berdasarkan input yang diberikan.

Contoh sederhana:

Input = 2

Weight = 3

Prediction = Input × Weight

Maka:

Prediction = 2 × 3 = 6

Dalam neural network, prediction dilakukan menggunakan operasi matematika terhadap input dan bobot.

Bobot menentukan seberapa besar pengaruh suatu input terhadap hasil prediksi.

## 2.3 Weight

Weight atau bobot merupakan parameter yang menentukan pengaruh input terhadap output.

Misalnya:

Input = 2

Weight = 3

Output = 2 × 3 = 6

Jika weight berubah menjadi 4:

Output = 2 × 4 = 8

Artinya, perubahan weight dapat mengubah hasil prediksi.

Dalam proses training, model akan menyesuaikan nilai weight agar prediksi semakin mendekati target.

## 2.4 Error

Error merupakan selisih antara hasil prediksi model dengan nilai target sebenarnya.

Contoh:

Target = 10

Prediction = 6

Error = Target - Prediction

Error = 10 - 6 = 4

Error digunakan untuk mengetahui seberapa jauh hasil prediksi dari target.

Semakin kecil error, semakin dekat prediksi model dengan nilai yang diharapkan.

## 2.5 Learning Process

Proses pembelajaran pada model dapat dijelaskan melalui beberapa langkah:

1. Model menerima input.
2. Model menghasilkan prediction.
3. Prediction dibandingkan dengan target.
4. Model menghitung error.
5. Weight diperbarui berdasarkan error.
6. Proses diulang sampai model menghasilkan prediksi yang lebih baik.

Proses ini merupakan dasar dari training pada neural network.

## 2.6 Types of Machine Learning

Secara umum, Machine Learning memiliki beberapa pendekatan:

### 1. Supervised Learning

Model belajar menggunakan data yang memiliki label atau target.

Contoh:
- Prediksi harga rumah.
- Klasifikasi email spam.
- Prediksi kelulusan mahasiswa.

### 2. Unsupervised Learning

Model belajar dari data tanpa label untuk menemukan pola atau kelompok.

Contoh:
- Clustering pelanggan.
- Pengelompokan dokumen.

### 3. Reinforcement Learning

Model belajar melalui interaksi dengan lingkungan menggunakan reward dan penalty.

Contoh:
- Robot yang belajar berjalan.
- AI yang belajar bermain game.

## 3.1 Implementing a Simple Prediction

Pada bagian ini, kita akan membuat prediksi sederhana menggunakan Python.

Model menggunakan satu input dan satu weight.

Rumus:

Prediction = Input × Weight

Tujuannya adalah memahami bahwa weight memiliki pengaruh langsung terhadap hasil prediksi.

```python
# Data input
input_value = 2

# Bobot model
weight = 3

# Menghitung prediksi
prediction = input_value * weight

print("Input:", input_value)
print("Weight:", weight)
print("Prediction:", prediction)
```

## 3.2 Code Explanation

Kode di atas merupakan bentuk paling sederhana dari proses prediction.

- `input_value` merupakan data yang diberikan kepada model.
- `weight` merupakan parameter yang menentukan pengaruh input.
- `prediction` merupakan hasil perkalian antara input dan weight.

Jika nilai weight berubah, hasil prediction juga akan berubah.

Konsep ini akan dikembangkan menjadi neural prediction pada chapter berikutnya.

## 4.1 Calculating Prediction Error

Setelah menghasilkan prediksi, kita perlu membandingkannya dengan target sebenarnya.

Rumus:

Error = Target - Prediction

Contoh:
- Target = 10
- Prediction = 6
- Error = 4

Error menunjukkan seberapa jauh hasil prediksi dari target.

```python
# Nilai target sebenarnya
target = 10

# Hasil prediksi model
prediction = 6

# Menghitung error
error = target - prediction

print("Target:", target)
print("Prediction:", prediction)
print("Error:", error)
```

## 4.2 Code Explanation

Pada kode tersebut:

- `target` menyimpan nilai yang seharusnya diprediksi.
- `prediction` menyimpan hasil prediksi model.
- `error` menghitung selisih antara target dan prediction.

Jika error bernilai positif, prediction lebih kecil daripada target.

Jika error bernilai negatif, prediction lebih besar daripada target.

Nilai error akan digunakan untuk memahami bagaimana model memperbaiki bobotnya.
