# Chapter 3 — Introduction to Neural Prediction: Forward Propagation

## 1.1 Overview

Chapter ini mulai membahas bagaimana neural network menghasilkan prediksi melalui proses yang disebut Forward Propagation.

Forward Propagation adalah proses mengalirkan data dari input menuju output melalui operasi matematika menggunakan weight.

Pada chapter ini, kita akan mempelajari:
- Cara kerja neural prediction.
- Hubungan input dan weight.
- Perkalian input dengan weight.
- Penggunaan beberapa input.
- Dot product untuk menghitung prediksi.

## 2.1 What is Forward Propagation?

Forward Propagation adalah proses menghitung output neural network berdasarkan input dan parameter yang dimiliki model.

Secara sederhana:

Input → Weight → Prediction

Input merupakan data yang diberikan kepada model.

Weight merupakan nilai yang menentukan pengaruh setiap input.

Prediction merupakan hasil perhitungan model.

Pada neural network sederhana dengan satu input dan satu weight, rumusnya:

Prediction = Input × Weight

Proses ini disebut forward karena perhitungan berjalan dari input menuju output.

## 2.2 Neural Prediction with One Input

Pada neural network sederhana, satu input dapat dikalikan dengan satu weight untuk menghasilkan prediksi.

Contoh:

Input = 2

Weight = 3

Prediction = 2 × 3 = 6

Jika weight berubah, hasil prediksi juga berubah.

Hal ini menunjukkan bahwa weight merupakan bagian penting yang akan dipelajari dan diperbarui selama training.

## 2.3 Neural Prediction with Multiple Inputs

Pada kasus yang lebih kompleks, model dapat menerima lebih dari satu input.

Setiap input memiliki weight masing-masing.

Contoh:

Input = [2, 3, 4]

Weight = [0.1, 0.2, 0.3]

Prediction dihitung dengan menjumlahkan hasil perkalian setiap input dengan weight-nya.

Rumus:

Prediction = (Input1 × Weight1) + (Input2 × Weight2) + (Input3 × Weight3)

Dengan demikian, setiap input memberikan kontribusi terhadap hasil prediksi.

## 2.4 Dot Product

Dot product merupakan operasi matematika yang digunakan untuk mengalikan elemen-elemen dari dua array, kemudian menjumlahkan hasilnya.

Contoh:

Input = [2, 3, 4]

Weight = [0.1, 0.2, 0.3]

Dot Product:

= (2 × 0.1) + (3 × 0.2) + (4 × 0.3)

= 0.2 + 0.6 + 1.2

= 2.0

Dalam neural network, dot product digunakan untuk menghitung kombinasi input dan weight secara efisien.

## 3.1 Forward Propagation with One Input

Pada implementasi pertama, kita akan membuat neural prediction menggunakan satu input dan satu weight.

Rumus:

Prediction = Input × Weight

Tujuannya adalah memahami bagaimana sebuah neuron menghasilkan output dari input yang diberikan.

```python
# Input data
input_value = 2

# Weight model
weight = 3

# Forward propagation
prediction = input_value * weight

print("Input:", input_value)
print("Weight:", weight)
print("Prediction:", prediction)
```

## 3.2 Code Explanation

Kode tersebut melakukan forward propagation sederhana.

- `input_value` menyimpan data input.
- `weight` menyimpan bobot model.
- `prediction` merupakan hasil perkalian input dan weight.

Hasil prediksi akan berubah jika nilai input atau weight diubah.

Pada tahap ini, weight masih ditentukan secara manual dan belum dipelajari melalui training.

## 4.1 Multiple Inputs and Weights

Selanjutnya, kita akan menggunakan beberapa input dan weight.

Setiap input dikalikan dengan weight yang sesuai, kemudian seluruh hasilnya dijumlahkan.

Rumus:

Prediction = Σ(Input × Weight)

Simbol Σ berarti menjumlahkan seluruh hasil perkalian.

Implementasi ini menunjukkan bagaimana satu neuron dapat menerima beberapa input sekaligus.

```python
import numpy as np

# Beberapa input
inputs = np.array([2, 3, 4])

# Weight untuk setiap input
weights = np.array([0.1, 0.2, 0.3])

# Forward propagation menggunakan dot product
prediction = np.dot(inputs, weights)

print("Inputs:", inputs)
print("Weights:", weights)
print("Prediction:", prediction)
```

## 4.2 Code Explanation

Penjelasan kode:

- `np.array()` digunakan untuk membuat array input dan weight.
- `np.dot()` melakukan operasi dot product.
- Setiap input dikalikan dengan weight yang bersesuaian.
- Seluruh hasil perkalian dijumlahkan menjadi satu nilai prediction.

Dengan menggunakan NumPy, kita tidak perlu melakukan perkalian dan penjumlahan secara manual.

## 5.1 Comparing Manual Calculation and Dot Product

Untuk memastikan hasil perhitungan, kita dapat membandingkan dot product dengan perhitungan manual menggunakan loop.

Kedua cara tersebut seharusnya menghasilkan nilai yang sama.

```python
import numpy as np

inputs = np.array([2, 3, 4])
weights = np.array([0.1, 0.2, 0.3])

# Perhitungan manual
manual_prediction = 0

for i in range(len(inputs)):
    manual_prediction += inputs[i] * weights[i]

# Perhitungan menggunakan NumPy
numpy_prediction = np.dot(inputs, weights)

print("Manual Prediction:", manual_prediction)
print("NumPy Prediction:", numpy_prediction)
```

## 5.2 Code Explanation

Pada perhitungan manual, loop digunakan untuk mengalikan setiap input dengan weight yang sesuai.

Hasil perkalian kemudian dijumlahkan ke dalam variabel `manual_prediction`.

Sementara itu, `np.dot()` melakukan operasi yang sama secara langsung.

Kedua pendekatan menghasilkan nilai yang sama, tetapi NumPy lebih ringkas dan praktis untuk operasi array.

## 6.1 The Effect of Changing Weights

Weight memiliki pengaruh langsung terhadap hasil prediction.

Untuk melihat pengaruhnya, kita akan menggunakan input yang sama tetapi mengubah nilai weight.

Jika weight berubah, prediction juga dapat berubah meskipun input tidak berubah.

```python
import numpy as np

inputs = np.array([2, 3, 4])

# Weight pertama
weights_1 = np.array([0.1, 0.2, 0.3])

# Weight kedua
weights_2 = np.array([0.2, 0.3, 0.4])

# Menghitung prediction
prediction_1 = np.dot(inputs, weights_1)
prediction_2 = np.dot(inputs, weights_2)

print("Prediction 1:", prediction_1)
print("Prediction 2:", prediction_2)
```

## 6.2 Code Explanation

Pada kode tersebut, input dibuat sama untuk kedua percobaan.

Perbedaannya hanya terletak pada nilai weight.

Hasilnya menunjukkan bahwa perubahan weight dapat mengubah prediction.

Inilah alasan mengapa proses training diperlukan, yaitu untuk menemukan nilai weight yang menghasilkan prediksi lebih baik.
