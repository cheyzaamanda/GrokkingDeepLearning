## 1. Introduction to PyTorch

PyTorch is a Python library for tensor computation and deep learning.

A tensor is a multidimensional data structure similar to a NumPy array. PyTorch tensors can support automatic differentiation, which is useful for calculating gradients during model training.

The `requires_grad=True` setting tells PyTorch to track operations involving a tensor so that gradients can be calculated using `backward()`.

```python
import torch

# Create tensors
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print("Tensor a:", a)
print("Tensor b:", b)

print("Addition:", a + b)
print("Element-wise multiplication:", a * b)
print("Dot product:", torch.dot(a, b))
```

```python
# Create a 2D tensor
matrix = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

print("Matrix:")
print(matrix)

print("\nShape:", matrix.shape)
print("Transpose:")
print(matrix.T)

print("\nMatrix multiplied by itself:")
print(torch.matmul(matrix, matrix))
```

```python
# Create a tensor whose gradient will be tracked
x = torch.tensor(3.0, requires_grad=True)

# Define a mathematical function: y = x^2 + 2x
y = x**2 + 2*x

# Calculate the gradient
y.backward()

print("x:", x.item())
print("y:", y.item())
print("Gradient dy/dx:", x.grad.item())
```

## 2. Automatic Differentiation

In the previous example:

y = x² + 2x

The derivative is:

dy/dx = 2x + 2

When x = 3:

dy/dx = 2(3) + 2 = 8

PyTorch calculates this gradient automatically when `backward()` is called.

Automatic differentiation reduces the need to manually derive and implement gradients for every operation in a complex neural network.

```python
import numpy as np
import torch

# NumPy array
numpy_array = np.array([1.0, 2.0, 3.0])

# Convert NumPy array to PyTorch tensor
torch_tensor = torch.from_numpy(numpy_array)

print("NumPy array:", numpy_array)
print("PyTorch tensor:", torch_tensor)

# Convert tensor back to NumPy
converted_array = torch_tensor.numpy()
print("Converted array:", converted_array)
```
